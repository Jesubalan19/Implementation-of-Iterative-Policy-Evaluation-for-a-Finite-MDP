```python
import gymnasium as gym
import numpy as np
```

```python
# Create FrozenLake environment
env = gym.make("FrozenLake-v1", map_name="4x4", is_slippery=True)
```

```python
# Access the unwrapped environment to use the transition model
env = env.unwrapped
```

```python
# Number of states and actions
n_states = env.observation_space.n
n_actions = env.action_space.n
```

```python
# Parameters
gamma = 0.99
theta = 1e-8
```

```python
# Random policy: each action has equal probability
policy = np.ones((n_states, n_actions)) / n_actions
```

```python
# Initialize value function
V = np.zeros(n_states)
# -------------------------------------------------
# Policy Evaluation Function
# -------------------------------------------------

def policy_evaluation(env, policy, gamma=0.99, theta=1e-8):

    V = np.zeros(env.observation_space.n)
    iteration = 0

    while True:
        delta = 0
        new_V = np.copy(V)

        for s in range(env.observation_space.n):
            v = 0

            for a, action_prob in enumerate(policy[s]):

                for prob, next_state, reward, terminated in env.P[s][a]:
                    v += action_prob * prob * (
                        reward + gamma * V[next_state] * (not terminated)
                    )

            new_V[s] = v
            delta = max(delta, abs(V[s] - new_V[s]))

        V = new_V
        iteration += 1
        if delta < theta:
            break

    return V, iteration
```

```python

# Run policy evaluation
V, iterations = policy_evaluation(env, policy, gamma, theta)

print("Name: Jesubalan A")
print("Register Number: 212223240060")
print("Number of iterations:", iterations)
print("\nState-Value Function:")
print(V)

print("Name: Jesubalan A")
print("Register Number: 212223240060")
print("\nState-Value Function as 4x4 Grid:")
print(np.round(V.reshape(4, 4), 4))

env.close()
```
